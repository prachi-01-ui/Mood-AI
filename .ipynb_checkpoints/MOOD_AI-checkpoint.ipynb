```python
from gtts import gTTS
import pygame
import os
import time
import uuid

pygame.mixer.init()

def speak(text):
    filename = f"voice_{uuid.uuid4().hex}.mp3"

    tts = gTTS(text=text, lang='en')
    tts.save(filename)

    pygame.mixer.music.load(filename)
    pygame.mixer.music.play()

    while pygame.mixer.music.get_busy():
        time.sleep(0.1)

    pygame.mixer.music.unload()

    try:
        os.remove(filename)
    except:
        pass
```

```python
def run_mood_ai():
    import math
    import time
    
    cap = cv2.VideoCapture(0)
    print("Press Q to capture mood")

    detected_mood = "neutral"

    # -------- LIVE DETECTION LOOP --------
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        result = DeepFace.analyze(frame, actions=['emotion'], enforce_detection=False)
        detected_mood = result[0]['dominant_emotion'].lower()

        cv2.putText(frame,
                    f"Mood Detected: {detected_mood}",
                    (30, 40),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    1,
                    (0, 255, 0),
                    2)

        cv2.imshow("Mood AI - Live Detection", frame)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    # Generate content after capture
    joke = random.choice(jokes.get(detected_mood, ["Stay positive!"]))
    poem = random.choice(poems.get(detected_mood, ["Keep going."]))
    story = random.choice(stories.get(detected_mood, ["You are stronger than you think."]))

    speak("Your detected mood is " + detected_mood)
    speak(joke)
    speak(poem)
    speak(story)

    # -------- FUTURISTIC DISPLAY LOOP --------
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        pulse = int((math.sin(time.time() * 3) + 1) * 50)

        theme_colors = {
            "happy": (0, 255, 0),
            "sad": (255, 120, 120),
            "angry": (0, 0, 255),
            "neutral": (0, 255, 255)
        }

        main_color = theme_colors.get(detected_mood, (255, 255, 255))

        glow_color = (
            min(main_color[0] + pulse, 255),
            min(main_color[1] + pulse, 255),
            min(main_color[2] + pulse, 255)
        )

        h, w, _ = frame.shape
        center = (w // 2, h // 2)

        # Glowing circular highlight
        cv2.circle(frame, center, 150, glow_color, 3)

        # Dark futuristic panel
        overlay = frame.copy()
        cv2.rectangle(overlay, (20, 20), (620, 380), (15, 15, 35), -1)
        cv2.addWeighted(overlay, 0.75, frame, 0.25, 0, frame)

        # Animated border
        cv2.rectangle(frame,
                      (5, 5),
                      (w - 5, h - 5),
                      glow_color,
                      3)

        y = 60

        cv2.putText(frame, "AI MOOD ANALYSIS",
                    (40, y),
                    cv2.FONT_HERSHEY_DUPLEX,
                    0.8,
                    glow_color,
                    2)
        y += 40

        sections = [
            f"Mood: {detected_mood}",
            f"Joke: {joke}",
            f"Poem: {poem}",
            f"Story: {story}"
        ]

        for line in sections:
            cv2.putText(frame,
                        line,
                        (50, y),
                        cv2.FONT_HERSHEY_SIMPLEX,
                        0.6,
                        (255, 255, 255),
                        1)
            y += 40

        cv2.putText(frame,
                    "Press M → Music | T → Talkbot | ESC → Exit",
                    (40, 350),
                    cv2.FONT_HERSHEY_SIMPLEX,
                    0.6,
                    (0, 255, 255),
                    2)

        cv2.imshow("Mood AI - Futuristic Mode", frame)

        key = cv2.waitKey(1)

        if key == ord('m'):
            play_music(detected_mood)

        if key == ord('t'):
            talkbot(detected_mood)

        if key == 27:  # ESC
            break

    cap.release()
    cv2.destroyAllWindows()
```

```python
def talkbot(mood):
    print("\nTalk to your AI Companion (type 'exit' to stop)\n")

    speak("I am here for you. How are you feeling?")

    while True:
        user_input = input("You: ")

        if user_input.lower() == "exit":
            speak("Take care. I am always here for you.")
            break

        if mood == "sad":
            response = "I understand. Would you like to share what made you feel this way?"
        elif mood == "happy":
            response = "That is wonderful! What made your day special?"
        elif mood == "angry":
            response = "Take a deep breath. Would you like to try a calming exercise?"
        else:
            response = "Tell me more about how you feel."

        print("AI:", response)
        speak(response)
```

```python
def play_music(mood):

    default_music = {
        "happy": "upbeat happy songs playlist",
        "sad": "calm relaxing instrumental music",
        "angry": "deep meditation calming music",
        "neutral": "lofi chill beats"
    }

    print("\n🎵 MUSIC OPTIONS")
    print("1 → Recommended music")
    print("2 → Custom search")
    print("3 → Skip")

    choice = input("Choose option (1/2/3): ")

    if choice == "1":
        query = default_music.get(mood, "relaxing music")
        print("Opening recommended music...")
        webbrowser.open(f"https://www.youtube.com/results?search_query={query}")

    elif choice == "2":
        custom = input("Enter your preferred music style: ")
        print("Opening custom music...")
        webbrowser.open(f"https://www.youtube.com/results?search_query={custom}")

    elif choice == "3":
        print("Music skipped.")
        return

    else:
        print("Invalid choice. No music opened.")
```
